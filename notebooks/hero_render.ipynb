# Hero render on free Colab GPU

This notebook is the GPU half of the hero-render contract defined in
`src/visuals/render_job.py`.

**Input.** It reads a `RenderJob` JSON document from a mounted Google Drive
folder (e.g. `MyDrive/videogen/jobs/<job_id>.json`). A `RenderJob` lists the
scenes to generate, each with a text `prompt` and a `duration_sec`, plus the
output `width`, `height`, and `fps`.

**Work.** For each scene it runs a text-to-video model over the scene prompt
and writes the resulting clip to the results folder.

**Output.** It writes a `RenderResult` JSON document back to the mounted Drive
folder (`MyDrive/videogen/results/<job_id>.json`). `RenderResult.outputs` maps
each scene `index` to the produced file name.

The cells below are a template: mount Drive, load the job, generate, and save
the result. The real model load/inference step is a placeholder.

```python
import json
from pathlib import Path

from google.colab import drive

drive.mount('/content/drive')

BASE = Path('/content/drive/MyDrive/videogen')
JOBS_DIR = BASE / 'jobs'
RESULTS_DIR = BASE / 'results'
RESULTS_DIR.mkdir(parents=True, exist_ok=True)

JOB_ID = 'demo-job'  # replace with the job id produced by the Oracle VM
job = json.loads((JOBS_DIR / f'{JOB_ID}.json').read_text())
print(job['model'], job['width'], 'x', job['height'], '@', job['fps'], 'fps')
print(len(job['scenes']), 'scene(s)')
```

```python
# Placeholder: load the text-to-video model on the GPU and render each scene.
outputs = {}
for scene in job['scenes']:
    # model.generate(scene['prompt'], duration_sec=scene['duration_sec'], ...)
    outputs[scene['index']] = f"scene_{scene['index']}.mp4"

result = {'job_id': job['job_id'], 'outputs': outputs}
out_path = RESULTS_DIR / f"{job['job_id']}.json"
out_path.write_text(json.dumps(result, indent=2, sort_keys=False))
print('wrote', out_path)
```
